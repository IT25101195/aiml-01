```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
from google.colab import drive
drive.mount('/content/drive')

df = pd.read_csv("/content/drive/My Drive/AIML-Project/G5/weatherHistory.csv")
print("Shape:", df.shape)
df.head()
```

```python
duplicates = df[df.duplicated()]
duplicates.shape
```

```python
df['Precip Type'] = df['Precip Type'].ffill() #ffill(), bfill(), mode() gives same values(assign to most rain)
```

```python
df.isnull().sum()
```

```python
df['Precip Type'].value_counts()
```

```python
df['Summary'].value_counts()
```

```python
df['Suitability'] = (df['Summary'] == 'Clear').astype(int)
```

```python
df = df.drop_duplicates()
```

```python
df = pd.get_dummies(
    df,
    columns=['Precip Type'],
    prefix='Precip',
    dtype=int
)
```

Handle Loud Cover

```python
df['Loud Cover'].value_counts()
```

```python
df = df.drop(columns=['Formatted Date', 'Summary', 'Daily Summary', 'Loud Cover'])
```

```python
df.head()
```

```python
df = df.rename(columns={
    'Temperature (C)': 'Temperature',
    'Apparent Temperature (C)': 'Apparent_Temperature',
    'Wind Speed (km/h)': 'Wind_Speed',
    'Wind Bearing (degrees)': 'Wind_Bearing',
    'Visibility (km)': 'Visibility',
    'Pressure (millibars)': 'Pressure',
    'Precip_rain': 'Precip_Rain',
    'Precip_snow': 'Precip_Snow'
})
```

Scaling the Dataset

```python
from sklearn.preprocessing import StandardScaler

num_cols = ['Temperature', 'Apparent_Temperature', 'Humidity', 'Wind_Speed', 'Wind_Bearing', 'Visibility', 'Pressure']
scaler = StandardScaler()
scaler.fit(df[num_cols])
```

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score
```

```python
X = df.drop(columns=['Suitability'])
y = df['Suitability']

pd_X = pd.DataFrame(X)
pd_y = pd.Series(y)

pd_X.head()
```

```python
pd_y.head()
```

```python
X_train, X_test, y_train, y_test = train_test_split(pd_X, pd_y, test_size=0.2, random_state=42)
print(f"Training Samples: {len(X_train)}, Testing Samples: {len(X_test)}")
```

```python
from sklearn.neighbors import KNeighborsClassifier
model_knn = KNeighborsClassifier(n_neighbors=5)
model_knn.fit(X_train, y_train)
```

```python
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report,
    confusion_matrix,
    roc_auc_score
)
y_pred_knn = model_knn.predict(X_test)
print("\nKNN Metrics:")
print(f"Accuracy: {accuracy_score(y_test, y_pred_knn) :.2f}")
print(f"Precision: {precision_score(y_test, y_pred_knn) :.2f}")
print(f"Recall: {recall_score(y_test, y_pred_knn) :.2f}")
print(f"F1 Score: {f1_score(y_test, y_pred_knn) :.2f}")
print(f"ROC AUC: {roc_auc_score(y_test, model_knn.predict_proba(X_test)[:, 1]) :.2f}")
```
